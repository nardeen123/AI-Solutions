In this project, a PowerBI dashboard was developed to monitor the financial performance of a healthcare institution. The users requested insights into total billing amounts, broken down by procedure, department, diagnosis, and service type, as well as a geographical representation of billing totals by city. Additionally, they wanted visibility into costs related to medication, treatments, and insurance. To enhance user experience, custom dropdown filters were implemented, allowing users to easily sort data by date and patient demographics, including race. In addition, darkmode option was also created with a button click option.
