```python
import gradio as gr
import requests

# Your API configuration
API_URL = "https://a8f815d5f4d4.ngrok-free.app"  # Your FastAPI server
API_ENDPOINT = "/eva-chat"  # Your text chat endpoint

class RAGClient:
    def __init__(self):
        self.api_url = API_URL
        self.endpoint = API_ENDPOINT

    def query(self, question):
        """
        Send query to your FastAPI RAG endpoint
        """
        try:
            # Prepare the request payload
            payload = {"query": question}

            # Make the API call to your FastAPI endpoint
            response = requests.post(
                f"{self.api_url}{self.endpoint}",
                json=payload,
                headers={"Content-Type": "application/json"},
                timeout=30
            )

            if response.status_code == 200:
                result = response.json()
                response_text = result.get("response", "No response received")

                # Add sources if available
                sources = result.get("sources", [])
                if sources:
                    response_text += f"\n\nSources: {', '.join(sources)}"

                return response_text
            else:
                error_detail = response.json().get("detail", "Unknown error")
                return f"Error: API returned status code {response.status_code}. Details: {error_detail}"

        except requests.exceptions.ConnectionError:
            return "❌ Error: Cannot connect to the API. Make sure your FastAPI server is running!"
        except requests.exceptions.Timeout:
            return "⏰ Error: Request timed out. The API is taking too long to respond."
        except Exception as e:
            return f"⚠️ Error: {str(e)}"

# Initialize client
rag_client = RAGClient()

# Create the Gradio interface
def chat_interface(message, history):
    """Function that handles the chat interaction"""
    if not message.strip():
        return "Please enter a question."

    # Get response from your RAG API
    response = rag_client.query(message)
    return response

# Create the interface
demo = gr.ChatInterface(
    fn=chat_interface,
    title="🏥 EVA Pharma Drug Information",
    description="Ask about drugs and medications at EVA Pharma",
    examples=[
        "Describe the dosage forms of alfacidocal",
        "What are the side effects for it?",
        "How to store it?",
        "What medications contain magnesium stearate?",
        "Describe the dosage forms of Anselacox"
    ],
    theme="soft"
)

# Launch the interface
if __name__ == "__main__":
    demo.launch(debug=True, share=True)
```
